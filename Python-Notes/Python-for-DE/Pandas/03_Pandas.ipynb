```python
def Quality(total_bill,tip):
    if total_bill/tip>0.25:
        return "Generous"
    else:
        return "Other"

Quality(16.09,2.09)
```

```python
New_df["Quantity"] = New_df[["total_bill","tip"]].apply(Quality)
```

```python
New_df["Quantity"] = New_df[["total_bill","tip"]].apply(lambda New_df : Quality(New_df["total_bill"], New_df["tip"]), axis = 1)
```
